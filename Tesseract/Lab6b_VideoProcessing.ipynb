```python
import cv2

def main():
    cap = cv2.VideoCapture(0, cv2.CAP_AVFOUNDATION)
    if not cap.isOpened():
        print("Error: Cannot access camera")
        return
    print("Camera started. Press 'q' to quit.")
    try:
        while True:
            ret, frame = cap.read()
            if not ret:
                print("Failed to grab frame")
                break
            cv2.imshow('Camera Feed', frame)
            if cv2.waitKey(10) & 0xFF == ord('q'):
                break
    finally:
        cap.release()
        cv2.destroyAllWindows()
        # Flush the Cocoa event queue so the window actually closes
        for _ in range(5):
            cv2.waitKey(1)
        print("Camera closed cleanly.")

if __name__ == "__main__":
    main()
```

```python
import cv2

def main():
    # Use Mac-friendly backend
    cap = cv2.VideoCapture(0, cv2.CAP_AVFOUNDATION)

    if not cap.isOpened():
        print("Error: Cannot access camera")
        return

    print("Camera started. Press 'q' to quit.")

    while True:
        ret, frame = cap.read()

        if not ret:
            print("Failed to grab frame")
            break

        cv2.imshow('Camera Feed', frame)

        # Press 'q' to exit
        if cv2.waitKey(10) & 0xFF == ord('q'):
            break

    # Clean shutdown (this works properly outside Jupyter)
    cap.release()
    cv2.destroyAllWindows()
    # Flush the Cocoa event queue so the window actually closes
    for _ in range(5):
        cv2.waitKey(1)
    print("Camera closed cleanly.")

if __name__ == "__main__":
    main()
```

### Saving Video to File
Again, this needs to be run locally.

```python
cap = cv2.VideoCapture(0)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

writer = cv2.VideoWriter('example_capture.mp4', cv2.VideoWriter_fourcc(*'XVID'),25, (width, height))

while True:
    ret, frame = cap.read()
    writer.write(frame)
    # cv2.imshow('frame',frame)
    cv2.imshow('Saving this video to a file - Press ESC to exit',frame)
    
    # Press ESC to quit.
    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
writer.release()
cv2.destroyAllWindows()
# Flush the Cocoa event queue so the window actually closes
for _ in range(5):
    cv2.waitKey(1)
```

## Drawing on Video
In this section we will draw a rectangle on live webcam feed

```python
import cv2
# This function is based on a CV2 Event (Left button click)
def draw_rectangle(event,x,y,flags,param):
    global pt1,pt2,topLeft_clicked,botRight_clicked
    if event == cv2.EVENT_LBUTTONDOWN:
        if topLeft_clicked == True and botRight_clicked == True:
            topLeft_clicked = False
            botRight_clicked = False
            pt1 = (0,0)
            pt2 = (0,0)
        if topLeft_clicked == False:
            pt1 = (x,y)
            topLeft_clicked = True
        elif botRight_clicked == False:
            pt2 = (x,y)
            botRight_clicked = True

pt1 = (0,0)
pt2 = (0,0)
topLeft_clicked = False
botRight_clicked = False

cap = cv2.VideoCapture(0)
cv2.namedWindow('Drawing on Video - Press ESC to exit')
cv2.setMouseCallback('Drawing on Video - Press ESC to exit', draw_rectangle)


while True:
    ret, frame = cap.read()

    if topLeft_clicked:
        cv2.circle(frame, center=pt1, radius=5, color=(0,0,255), thickness=-1)

    #Draw rectangle
    if topLeft_clicked and botRight_clicked:
        cv2.rectangle(frame, pt1, pt2, (0, 0, 255), 2)
    cv2.imshow('Drawing on Video - Press ESC to exit', frame)

    # Press ESC to quit.
    if cv2.waitKey(25) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
